The VICTOR implementation here aims to get your transmission FTIR spectra processed. We remove degrees of freedom to simplify the process. Examples workflows working with these spectra can be found on the [ReadTheDocs](https://pyiroglass.readthedocs.io/en/latest/).

The files necessary include example transmission FTIR spectra, as well as 'Colab_Binder_ChemThick.CSV' file, containing all chemistry and thickness data. Find an example [here](https://github.com/sarahshi/PyIRoGlass/blob/main/Inputs/COLAB_BINDER/Colab_Binder_ChemThick.csv). Ensure that all of your transmission FTIR spectra names (before the .CSV) match up identically with the sample names in the chemistry and thickness file. This is necessary to ensure that the algorithm can match up samples.

# Load transmission FTIR spectra here:

```python
# Provide folder name containing transmission FTIR spectra
SPEC_PATH = "TransmissionSpectra"
```

# Load ChemThick.CSV here:

```python
# Provide CSV name for your chemistry and thickness file. Here, it is called Colab_Binder_ChemThick.csv. Modify it here for your file name.
CHEMTHICK_PATH = 'Colab_Binder_ChemThick.csv'
```

# Run PyIRoGlass here:

```python
# Set the output path name desired for your sample. We will call our output 'RESULTS'.
OUTPUT_PATH = 'RESULTS'

# Load a list of files and a dictionary of transmission FTIR spectra from the files identified earlier.
loader = pig.SampleDataLoader(SPEC_PATH, CHEMTHICK_PATH)
dfs_dict, chem, thick = loader.load_all_data()

# Run PyIRoGlass! df_output will be saved as a CSV called OUTPUT_PATH+'_DF.CSV' in the FINALDATA directory.
df_output, failures = pig.calculate_baselines(dfs_dict, OUTPUT_PATH)

# Default temperature and pressure conditions of analysis are set at 25 C and 1 bar. These are the defaults at LDEO.
# df_conc will be saved as a CSV called OUTPUT_PATH+'_H2OCO2.CSV' in the FINALDATA directory.
df_conc = pig.calculate_concentrations(df_output, chem, thick, OUTPUT_PATH)
```
